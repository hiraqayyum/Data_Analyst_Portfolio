```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
url='https://join.com/recruitment-hr-blog/best-job-sites-in-spain'

page= requests.get(url)
```

```python
soup = BeautifulSoup(page.text, 'html')
```

```python
tags= soup.find_all('h3')
print(tags)
```

```python
websites = [tag.text.strip() for tag in tags]
websites
```

```python
Top_websites = [site.split('. ',1)[1] for site in websites[0:12]]
Top_websites
```

```python
df = pd.DataFrame(Top_websites , columns=['Top Websites for Remote Jobs in Spain'])
df
```

```python
df.insert(0, 'Rank', range(1, len(df)+1))
```

```python
df
```

```python
df.rename(columns={"Rank": "Website Ranking"}, inplace=True)
```
